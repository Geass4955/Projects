```python
# 2. Clean the dataset (Pandas)
# Remove duplicate rows.
# Handle missing values (drop them or fill them, depending on the dataset).
# Verify that there are no missing values remaining.

# Libraries: Pandas
```

```python
# Remove duplicate rows
df.drop_duplicates(inplace=True)

# Fill missing values in text (object) columns
text_columns = [
    "Title_English",
    "Title_Native",
    "Format",
    "Genres",
    "Season",
    "Source",
    "Studios",
    "Authors_Staff"
]

for col in text_columns:
    df[col] = df[col].fillna("Unknown")

# Fill missing values in numerical columns
number_columns = [
    "Episodes",
    "Season_Year",
    "Average_Score",
    "Mean_Score"
]

for col in number_columns:
    df[col] = df[col].fillna(df[col].median())

# Verify there are no missing values
print(df.isnull().sum())
```

```python
# 3. Score analysis (Pandas + NumPy)
# Find the average, minimum, and maximum Average Score.
# Find the median Average Score.
# Calculate the standard deviation.
# Calculate the variance of the Average Score.

# Libraries: Pandas, NumPy
```

```python


# Average, Minimum, and Maximum Average Score
average_score = df["Average_Score"].mean()
minimum_score = df["Average_Score"].min()
maximum_score = df["Average_Score"].max()

print("Average Score:", average_score)
print("Minimum Score:", minimum_score)
print("Maximum Score:", maximum_score)

# Median Average Score
median_score = df["Average_Score"].median()
print("Median Score:", median_score)

# Standard Deviation
std_score = np.std(df["Average_Score"])
print("Standard Deviation:", std_score)

# Variance
variance_score = np.var(df["Average_Score"])
print("Variance:", variance_score)
```

```python
# 19. Average score by season and format (Pandas + Seaborn)
# Create a pivot table with Season as rows and Format as columns.
# Fill the table with the average Average Score.
# Display the result using a heatmap.

# Libraries: Pandas, Seaborn
```

```python
pivot_table = df.pivot_table(
    values="Average_Score",
    index="Season",
    columns="Format",
    aggfunc="mean"
)

print(pivot_table)
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    pivot_table,
    annot=True,
    cmap="YlGnBu",
    fmt=".1f"
)

plt.title("Average Score by Season and Format")
plt.xlabel("Format")
plt.ylabel("Season")

plt.show()
```
