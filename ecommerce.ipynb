```python
#10 Time Series Analysis (Pandas)
# Monthly sales.
# Monthly number of orders.
# Best sales month.
# Worst sales month.
```

```python
df["order_date"] = pd.to_datetime(df["order_date"])
```

```python
# Create a Month-Year column
df["Month"] = df["order_date"].dt.to_period("M")

# Calculate monthly sales
monthly_sales = df.groupby("Month")["order_value_usd"].sum()

print(monthly_sales)
```

```python
monthly_orders = df.groupby("Month")["order_id"].count()

print(monthly_orders)
```

```python
best_month = monthly_sales.idxmax()
best_sales = monthly_sales.max()

print("Best Sales Month:", best_month)
print("Total Sales:", best_sales)
```

```python
worst_month = monthly_sales.idxmin()
worst_sales = monthly_sales.min()

print("Worst Sales Month:", worst_month)
print("Total Sales:", worst_sales)
```

```python
# #12 Data Visualization (Matplotlib & Seaborn)

# Create:

# Bar chart of product categories
# Pie chart of payment methods
# Histogram of order values
# Boxplot of customer ratings
# Countplot of returned orders
# Scatterplot: Delivery Time vs Rating
# Line chart of monthly sales
# Heatmap of feature correlations
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=df,
              x="product_category")

plt.title("Number of Orders by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)

plt.show()
```

```python
payment_counts = df["payment_method"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(payment_counts,
        labels=payment_counts.index,
        autopct="%1.1f%%",
        startangle=90)

plt.title("Payment Method Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(data=df,
             x="order_value_usd",
             bins=20,
             kde=True)

plt.title("Distribution of Order Values")
plt.xlabel("Order Value (USD)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.boxplot(data=df,
            y="customer_rating")

plt.title("Customer Rating Distribution")
plt.ylabel("Customer Rating")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(data=df,
              x="returned")

plt.title("Returned Orders")
plt.xlabel("Returned")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df,
                x="delivery_time_days",
                y="customer_rating")

plt.title("Delivery Time vs Customer Rating")
plt.xlabel("Delivery Time (Days)")
plt.ylabel("Customer Rating")

plt.show()
```

```python
monthly_sales = df.groupby("Month")["order_value_usd"].sum()

plt.figure(figsize=(10,5))

plt.plot(monthly_sales.index.astype(str),
         monthly_sales.values,
         marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales (USD)")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,6))

correlation = df.corr(numeric_only=True)

sns.heatmap(correlation,
            annot=True,
            cmap="coolwarm")

plt.title("Feature Correlation Heatmap")

plt.show()
```
