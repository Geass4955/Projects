```python
#2 Data Cleaning
# Remove duplicates.
# Handle missing values.
# Convert purchase and sold dates into datetime.
# Verify cleaned data.
```

```python
df["Sold Date"] = pd.to_datetime(df["Sold Date"])
```

```python
#10 Date Analysis
# Cars purchased by year.
# Cars sold by year.
# Monthly sales trend.
```

```python
purchased_by_year = df.groupby(df["Purchased Date"].dt.year)["ID"].count()

print(purchased_by_year)
```

```python
sold_by_year = df.groupby(df["Sold Date"].dt.year)["ID"].count()

print(sold_by_year)
```

```python
df["Sale Month"] = df["Sold Date"].dt.to_period("M")
```

```python
monthly_sales = df.groupby("Sale Month")["Sold Price-$"].sum()

print(monthly_sales)
```

```python
# Create Month-Year column

df["Sale Month"] = df["Sold Date"].dt.strftime("%b %Y")
```

```python
monthly_sales = df.groupby("Sale Month")["Sold Price-$"].sum()

print(monthly_sales)
```

```python
#12 Visualization

# Create:

# Manufacturer bar chart
# Price histogram
# Mileage histogram
# Boxplot of prices
# Sales status countplot
# Scatterplot Mileage vs Price
# Scatterplot Engine Power vs Price
# Heatmap correlation
```

```python
plt.figure(figsize=(12,6))

sns.countplot(
    data=df,
    x="Manufacturer Name",
    order=df["Manufacturer Name"].value_counts().index
)

plt.title("Number of Cars by Manufacturer")
plt.xlabel("Manufacturer")
plt.ylabel("Number of Cars")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="Sold Price-$",
    bins=20,
    kde=True
)

plt.title("Distribution of Selling Prices")
plt.xlabel("Selling Price ($)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="Mileage-KM",
    bins=20,
    kde=True
)

plt.title("Distribution of Mileage")
plt.xlabel("Mileage (KM)")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.boxplot(
    data=df,
    y="Sold Price-$"
)

plt.title("Boxplot of Selling Prices")
plt.ylabel("Selling Price ($)")

plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.countplot(
    data=df,
    x="Car Sale Status"
)

plt.title("Car Sale Status")
plt.xlabel("Sale Status")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Mileage-KM",
    y="Sold Price-$"
)

plt.title("Mileage vs Selling Price")
plt.xlabel("Mileage (KM)")
plt.ylabel("Selling Price ($)")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="Engine Power-HP",
    y="Sold Price-$"
)

plt.title("Engine Power vs Selling Price")
plt.xlabel("Engine Power (HP)")
plt.ylabel("Selling Price ($)")

plt.show()
```

```python
plt.figure(figsize=(10,8))

correlation = df.corr(numeric_only=True)

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.boxplot(
    data=df,
    x="Manufacturer Name",
    y="Sold Price-$"
)

plt.xticks(rotation=45)
plt.title("Selling Price by Manufacturer")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    data=df,
    x="Car Type",
    y="Sold Price-$"
)

plt.xticks(rotation=45)
plt.title("Average Selling Price by Car Type")

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.barplot(
    data=df,
    x="Manufacturer Name",
    y="Margin-%"
)

plt.xticks(rotation=45)
plt.title("Average Margin by Manufacturer")

plt.show()
```
