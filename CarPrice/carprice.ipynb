```python
# 2. Basic statistics (Pandas + NumPy)
# Find the average, minimum, and maximum car price.
# Calculate the average mileage.
# Find the median and standard deviation of price.

# Libraries: Pandas, NumPy
```

```python
# Find the average, minimum, and maximum car price.

Average_price = df["Price"].mean()
Minimum_price = df["Price"].min()
Maximum_price = df["Price"].max()

print("Average Price:", Average_price)
print("Minimum Price:", Minimum_price)
print("Maximum Price:", Maximum_price)
```

```python
# Calculate the average mileage.

Average_mileage = df["Mileage"].mean()
print("Average Mileage:", Average_mileage)
```

```python
# Find the median and standard deviation of price.


Median_price = np.median(df["Price"])
Std_price = np.std(df["Price"])

print("Median Price:", Median_price)
print("Standard Deviation of Price:", Std_price)
```

```python
# 3. Brand analysis (Pandas)
# Count how many cars belong to each brand.
# Find the average price of each brand.
# Display the top 5 most expensive brands.

# Libraries: Pandas
```

```python
Brand_count = df["Brand"].value_counts()
print(Brand_count)
```

```python
Average_price = df.groupby("Brand")["Price"].mean()
print(Average_price)
```

```python
Top_5_brands = df.groupby("Brand")["Price"].mean().sort_values(ascending=False).head(5)
print(Top_5_brands)
```

```python
# 4. Fuel type analysis (Pandas + Seaborn)
# Count the number of cars by fuel type.
# Create a count plot of fuel types.

# Libraries: Pandas, Seaborn
```

```python
Fuel_count = df["Fuel Type"].value_counts()
print(Fuel_count)
```

```python
sns.countplot(data=df, x="Fuel Type")
plt.title("Number of Cars by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Number of Cars")
plt.show()
```

```python
# 5. Price vs Mileage (Matplotlib)
# Create a scatter plot between Mileage and Price.
# Observe whether mileage affects price.

# Libraries: Matplotlib
```

```python


plt.figure(figsize=(8,5))

plt.scatter(df["Mileage"], df["Price"])

plt.title("Price vs Mileage")
plt.xlabel("Mileage")
plt.ylabel("Price")

plt.show()
```

```python
plt.scatter(df["Mileage"], df["Price"], color="purple")

plt.xlabel("Mileage", color="blue")
plt.ylabel("Price", color="red")

plt.title("Price vs Mileage")
plt.show()
```

```python
# 6. Distribution of car prices (Seaborn)
# Plot a histogram with KDE for Price.

# Libraries: Seaborn
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="Price",
    bins=20,
    kde=True,
    color="skyblue",
    edgecolor="black"
)

plt.title("Distribution of Car Prices")
plt.xlabel("Price")
plt.ylabel("Number of Cars")

plt.show()
```

```python
# 7. Correlation between numerical columns (Pandas + Seaborn)
# Calculate the correlation matrix.
# Display it using a heatmap.

# Libraries: Pandas, Seaborn
```

```python
correlation = df.corr(numeric_only=True)

print(correlation)
```

```python
plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Matrix")
plt.show()
```

```python
# 8. Compare the distributions of important numerical features (Seaborn)

# Create a figure with three subplots to compare the distributions of:

# Car Price
# Mileage
# Engine Size
# Libraries: Seaborn, Matplotlib
```

```python


plt.figure(figsize=(15, 5))

# Car Price
plt.subplot(1, 3, 1)
sns.histplot(data=df, x="Price", kde=True, color="skyblue")
plt.title("Car Price Distribution")

# Mileage
plt.subplot(1, 3, 2)
sns.histplot(data=df, x="Mileage", kde=True, color="lightgreen")
plt.title("Mileage Distribution")

# Engine Size
plt.subplot(1, 3, 3)
sns.histplot(data=df, x="Engine Size", kde=True, color="salmon")
plt.title("Engine Size Distribution")

plt.tight_layout()
plt.show()
```

```python
sns.pairplot(
    df,
    vars=["Price", "Mileage", "Engine Size"],
    diag_kind="hist"
)
plt.show()
```

```python
# 9. Compare price based on different categories (Seaborn)

# Create a figure with three boxplots:

# Price by Fuel Type
# Price by Transmission
# Price by Seller Type (or Owner, depending on your dataset)

# Libraries: Seaborn, Matplotlib
```

```python

plt.figure(figsize=(18,5))

# Price by Fuel Type
plt.subplot(1,3,1)
sns.boxplot(data=df, x="Fuel Type", y="Price", color="skyblue")
plt.title("Price by Fuel Type")

# Price by Transmission
plt.subplot(1,3,2)
sns.boxplot(data=df, x="Transmission", y="Price", color="lightgreen")
plt.title("Price by Transmission")

# Price by Condition
plt.subplot(1,3,3)
sns.boxplot(data=df, x="Condition", y="Price", color="salmon")
plt.title("Price by Condition")

plt.tight_layout()
plt.show()
```

```python
# 10. Compare relationships with Price (Seaborn)

# Create a figure with three scatter plots:

# Mileage vs Price
# Engine Size vs Price
# Horsepower (or Max Power) vs Price

# Use Fuel Type as the hue in all three graphs.

# Libraries: Seaborn, Matplotlib
```

```python


plt.figure(figsize=(24, 8))

# Mileage vs Price
plt.subplot(1, 3, 1)
sns.scatterplot(data=df, x="Mileage", y="Price", hue="Fuel Type")
plt.title("Mileage vs Price", fontsize=16)
plt.xlabel("Mileage", fontsize=13)
plt.ylabel("Price", fontsize=13)

# Engine Size vs Price
plt.subplot(1, 3, 2)
sns.scatterplot(data=df, x="Engine Size", y="Price", hue="Fuel Type", legend=False)
plt.title("Engine Size vs Price", fontsize=16)
plt.xlabel("Engine Size", fontsize=13)
plt.ylabel("Price", fontsize=13)

# Year vs Price
plt.subplot(1, 3, 3)
sns.scatterplot(data=df, x="Year", y="Price", hue="Fuel Type", legend=False)
plt.title("Year vs Price", fontsize=16)
plt.xlabel("Year", fontsize=13)
plt.ylabel("Price", fontsize=13)

plt.tight_layout()
plt.show()
```
